# 🎧 **Řešení projektu: Klasifikace hudebních žánrů podle audio vlastností písní (Spotify)**

## 1️⃣ Výběr datasetu

Pro projekt je zvolen dataset **[Spotify Songs](https://github.com/rfordatascience/tidytuesday/tree/main/data/2020/2020-01-21)** z komunitního projektu *TidyTuesday*. Data pocházejí ze **Spotify Web API** (stažená pomocí R balíčku `spotifyr`) a obsahují písně z playlistů šesti hlavních hudebních žánrů: **EDM, Latin, Pop, R&B, Rap a Rock**.

🎯 **Cíl:** na základě **audio vlastností** písně (taneční potenciál, energie, hlasitost, tempo, podíl mluveného slova, …) **předpovědět její žánr** → **vícetřídní klasifikace** se 6 třídami.

**Proč tento dataset:**
- reálná data z reálné služby, ne učebnicový příklad,
- všechny vstupní vlastnosti jsou číselné → vhodné pro rozhodovací stromy, SVM i neuronové sítě,
- žánry se přirozeně překrývají (např. pop × latin × r&b), takže úloha není triviální a ve výsledcích je o čem diskutovat.

**Splnění doporučené velikosti datasetu** (tabulka *Tabulární regrese / klasifikace* v zadání):

| Modelový typ | Doporučeno | Spotify: 28 356 unikátních písní, F = 12 |
|---|---|---|
| Lineární / logistická regrese | ≥ 10–20 × F, min. 500–2 000 | ✔ |
| Rozhodovací stromy / SVM | 2 000 – 10 000 | ✔ (dat je víc → pro ladění SVM lze použít náhodný vzorek) |
| Malé NN (dense sítě) | 50–200 × F, min. 1 000 | ✔ |
| Typický počet featur | 5–100 | ✔ (12 audio vlastností) |

## 2️⃣ Popis datasetu

Každý řádek = jedna píseň v jednom playlistu. Dataset má **32 833 řádků a 23 sloupců**, které lze rozdělit do tří skupin.

### 🏷️ Identifikátory a metadata (do modelu nevstupují)

| Sloupec | Typ | Význam |
|---|---|---|
| `track_id` | text | unikátní ID písně na Spotify |
| `track_name` | text | název písně |
| `track_artist` | text | interpret |
| `track_popularity` | číslo 0–100 | popularita písně na Spotify (vyšší = populárnější) |
| `track_album_id` | text | unikátní ID alba |
| `track_album_name` | text | název alba |
| `track_album_release_date` | text (datum) | datum vydání alba |
| `playlist_name` | text | název playlistu, ze kterého píseň pochází |
| `playlist_id` | text | ID playlistu |
| `playlist_subgenre` | text | podžánr playlistu (např. *dance pop*, *trap*, *hard rock*) |

### 🎯 Cílová proměnná (`y`)

| Sloupec | Typ | Význam |
|---|---|---|
| `playlist_genre` | kategorie | žánr playlistu: `edm`, `latin`, `pop`, `r&b`, `rap` nebo `rock` |

### 🎚️ Audio vlastnosti – vstupy modelu (`X`)

Hodnoty spočítalo Spotify automatickou analýzou zvukové stopy.

| Sloupec | Rozsah | Význam |
|---|---|---|
| `danceability` | 0–1 | jak moc se píseň hodí k tanci (tempo, stabilita rytmu, síla beatu) |
| `energy` | 0–1 | vnímaná intenzita a aktivita – energické písně jsou rychlé, hlasité a „hlučné“ |
| `key` | 0–11 | tónina (0 = C, 1 = C♯/D♭, 2 = D, …; -1 = nerozpoznáno) |
| `loudness` | cca -60–0 dB | průměrná hlasitost skladby v decibelech |
| `mode` | 0 / 1 | stupnice: 1 = dur, 0 = moll |
| `speechiness` | 0–1 | podíl mluveného slova (> 0,66 téměř jen řeč, 0,33–0,66 hudba s řečí, např. rap) |
| `acousticness` | 0–1 | míra jistoty, že je skladba akustická |
| `instrumentalness` | 0–1 | pravděpodobnost, že skladba neobsahuje zpěv (> 0,5 = instrumentální) |
| `liveness` | 0–1 | pravděpodobnost živé nahrávky s publikem (> 0,8 = téměř jistě live) |
| `valence` | 0–1 | „pozitivita“ – vysoká = veselá, euforická; nízká = smutná, naštvaná |
| `tempo` | BPM | odhadované tempo v úderech za minutu |
| `duration_ms` | ms | délka skladby v milisekundách |

⚠️ **Na co si dát pozor:**
- `playlist_genre` je žánr **playlistu**, ne přímo písně. Stejná píseň se může objevit ve více playlistech, dokonce s různými žánry. Data proto obsahují **duplicity** (32 833 řádků, ale jen 28 356 unikátních `track_id`), které je potřeba vyřešit v rámci EDA.
- `track_popularity` není vlastnost zvuku, proto zatím mezi vstupy není – o jejím zařazení se rozhodne po EDA.
