## 1️⃣ Výběr datasetu

Pro projekt je zvolen dataset **[Spotify Songs](https://github.com/rfordatascience/tidytuesday/tree/main/data/2020/2020-01-21)** z komunitního projektu *TidyTuesday*. Data pocházejí ze **Spotify Web API** (stažená pomocí R balíčku `spotifyr`) a obsahují písně z playlistů šesti hlavních hudebních žánrů: **EDM, Latin, Pop, R&B, Rap a Rock**.

🎯 **Cíl:** na základě **audio vlastností** písně (taneční potenciál, energie, hlasitost, tempo, podíl mluveného slova, …) **předpovědět její žánr** → **vícetřídní klasifikace** se 6 třídami.

**Proč tento dataset:**
- reálná data z reálné služby, ne učebnicový příklad,
- všechny vstupní vlastnosti jsou číselné → vhodné pro rozhodovací stromy, SVM i neuronové sítě,

**Splnění doporučené velikosti datasetu** (tabulka *Tabulární regrese / klasifikace* v zadání):

| Modelový typ | Doporučeno | Spotify: 28 356 unikátních písní, F = 12 |

| Lineární / logistická regrese | ≥ 10–20 × F, min. 500–2 000 | ✔ |
| Rozhodovací stromy / SVM | 2 000 – 10 000 | ✔ (dat je víc → pro ladění SVM lze použít náhodný vzorek) |

| Typický počet featur | 5–100 | ✔ (12 audio vlastností) |

## 4️⃣ Explorační analýza dat (EDA)

Cílem EDA je data poznat dřív, než je předložíme modelu:

1. základní statistiky a chybějící hodnoty,
2. duplicity a vyčištění dat,
3. rozložení cílové proměnné (vyváženost tříd),
4. rozložení jednotlivých audio vlastností,
5. jak se vlastnosti liší mezi žánry,
6. korelace mezi vlastnostmi.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

TARGET = "playlist_genre"
AUDIO_FEATURES = [
    "danceability", "energy", "key", "loudness", "mode", "speechiness",
    "acousticness", "instrumentalness", "liveness", "valence", "tempo", "duration_ms",
]
CONTINUOUS = [c for c in AUDIO_FEATURES if c not in ("key", "mode")]
GENRES = sorted(df[TARGET].unique())
GENRE_COLORS = dict(zip(GENRES, sns.color_palette("Set2", len(GENRES))))
```

### 4.1 Základní statistiky a chybějící hodnoty

```python
df[AUDIO_FEATURES].describe().T.round(3)
```

- Všechny audio vlastnosti jsou kompletní (32 833 hodnot) a většina z nich leží v rozsahu 0–1.
- Vlastnosti mají **velmi rozdílná měřítka**: `duration_ms` je ve statisících, `tempo` ve stovkách a `loudness` je záporná (dB). Pro SVM a neuronové sítě bude proto nutné data **standardizovat**, rozhodovacím stromům to nevadí.
- Podezřelé je minimum `tempo = 0` a `duration_ms = 4000` (4 s) – podíváme se na ně při čištění.

```python
missing = df.isna().sum()
missing[missing > 0]
```

Chybí jen 5 názvů písní, interpretů a alb. Jde o metadata, která do modelu nevstupují, ale bez názvu a interpreta nejde řádek zkontrolovat na duplicity (viz další krok). Těchto 5 řádků proto při čištění vyřadíme.

### 4.2 Duplicity a vyčištění dat

Stejná píseň se v datech může objevit vícekrát:
- **se stejným `track_id`** – píseň je zařazená ve více playlistech,
- **s jiným `track_id`** – stejná skladba vydaná např. na singlu i na albu (stejný název i interpret).

Duplicity jsou problém ze dvou důvodů:
1. **Rozporný štítek** – pokud je stejná píseň v playlistech různých žánrů, nevíme, který žánr je „správný“, a model dostává protichůdné informace.
2. **Únik informací** – pokud se stejná píseň dostane do trénovacích i testovacích dat, model ji při testu jen „pozná“ a výsledek bude zkresleně dobrý.

```python
genres_per_track = df.groupby("track_id")[TARGET].nunique()

print(f"Řádků celkem:                         {len(df)}")
print(f"Unikátních track_id:                  {df['track_id'].nunique()}")
print(f"Unikátních písní (název + interpret): {df.groupby(['track_name', 'track_artist']).ngroups}")
print(f"track_id ve více různých žánrech:     {(genres_per_track > 1).sum()}")
```

```python
example_id = genres_per_track[genres_per_track > 1].index[0]
df.loc[df["track_id"] == example_id, ["track_name", "track_artist", "playlist_name", TARGET]]
```

Např. píseň *We Own It (Fast & Furious)* je současně v rapovém, R&B i EDM playlistu – pro určení žánru podle zvuku takový záznam nemá jednoznačnou správnou odpověď.

Kromě duplicit je v datech jedno **chybné měření** – 4sekundové intro, u kterého Spotify nenaměřilo téměř žádné hodnoty (tempo, danceability, valence… jsou 0):

```python
SONG = ["track_name", "track_artist"]
df.loc[df["tempo"] == 0, SONG + AUDIO_FEATURES]
```

**Postup čištění:**
1. vyřadit 5 řádků bez názvu/interpreta,
2. vyřadit písně, které se v datech vyskytují s více různými žánry (rozporný štítek),
3. z ostatních duplicit ponechat jen jeden výskyt každé písně (název + interpret),
4. vyřadit chybné měření s `tempo = 0`.

```python
df_clean = df.dropna(subset=SONG)
print(f"1) bez chybějícího názvu/interpreta: {len(df_clean)}")

genres_per_song = df_clean.groupby(SONG)[TARGET].transform("nunique")
df_clean = df_clean[genres_per_song == 1]
print(f"2) bez písní s více žánry:           {len(df_clean)}")

df_clean = df_clean.drop_duplicates(subset=SONG)
print(f"3) bez duplicit:                     {len(df_clean)}")

df_clean = df_clean[df_clean["tempo"] > 0].reset_index(drop=True)
print(f"4) bez chybného měření (tempo = 0):  {len(df_clean)}")

print(f"\nZ původních {len(df)} řádků zůstává {len(df_clean)} unikátních písní ({len(df_clean) / len(df):.0%}).")
```

Po vyčištění zůstává **24 341 unikátních písní** s jednoznačným žánrem (74 % původních řádků), což stále bohatě splňuje doporučenou velikost ze zadání.

Ostatní extrémní hodnoty jsou reálné, a proto je ponecháváme: 24 písní kratších než 1 minuta (intra, mezihry) a 6 písní s hlasitostí mírně nad 0 dB (velmi hlasitě masterované skladby).

### 4.3 Rozložení cílové proměnné

```python
genre_counts = df_clean[TARGET].value_counts().reindex(GENRES)

plt.figure(figsize=(9, 4.5))
plt.bar(genre_counts.index, genre_counts.values, color=[GENRE_COLORS[g] for g in GENRES])
for i, count in enumerate(genre_counts.values):
    plt.text(i, count + 60, f"{count}\n({count / len(df_clean):.1%})", ha="center", va="bottom")
plt.title("Počet písní podle žánru (po vyčištění)")
plt.xlabel("Žánr")
plt.ylabel("Počet písní")
plt.ylim(0, genre_counts.max() * 1.25)
plt.show()
```

Třídy jsou **přibližně vyvážené** (15–19 % na žánr), takže není potřeba data převzorkovávat a jako hlavní metriku můžeme použít přesnost (*accuracy*). Při dělení na trénovací a testovací data použijeme `stratify`, aby byl poměr žánrů v obou částech stejný.

Náhodné hádání by mělo přesnost zhruba 1/6 ≈ 17 % – to je spodní hranice, kterou musí model výrazně překonat.

### 4.4 Rozložení audio vlastností

```python
fig, axes = plt.subplots(3, 4, figsize=(18, 11))
for ax, col in zip(axes.flat, AUDIO_FEATURES):
    if col in ("key", "mode"):
        sns.histplot(df_clean[col], discrete=True, ax=ax)
    else:
        sns.histplot(df_clean[col], bins=40, ax=ax)
    ax.set_title(col)
    ax.set_xlabel("")
    ax.set_ylabel("")
fig.suptitle("Rozložení audio vlastností", fontsize=16)
plt.tight_layout()
plt.show()
```

- `danceability`, `valence` a `duration_ms` mají přibližně „zvonovité“ rozložení (typická píseň trvá 3–4 minuty), `energy` je posunutá k vyšším hodnotám – většina písní je spíše energická.
- `tempo` má **několik vrcholů** – nejvýraznější úzký vrchol je kolem 125–128 BPM. Jak ukáže další graf, jde hlavně o EDM, které má tempo velmi ustálené.
- `speechiness`, `acousticness`, `liveness` a hlavně `instrumentalness` jsou silně **zešikmené doprava**. U 75 % písní je `instrumentalness` prakticky nulová (< 0,01), tzn. naprostá většina písní má zpěv.
- `loudness` je naopak zešikmená doleva – polovina písní má hlasitost mezi −8,4 a −4,8 dB (moderní hlasitý mastering).
- `key` (tónina 0–11) je ve skutečnosti **kategorie**, ne číslo, a `mode` je binární (56 % písní v dur, 44 % v moll).

### 4.5 Audio vlastnosti podle žánru

Nejdůležitější část EDA: pokud se nějaká vlastnost mezi žánry výrazně liší, model ji může využít k jejich rozlišení. Krabicové grafy (boxploty) ukazují pro každý žánr medián (čára), prostřední polovinu hodnot (krabice) a rozptyl (vousy). Odlehlé body kvůli přehlednosti nezobrazujeme.

```python
fig, axes = plt.subplots(2, 5, figsize=(24, 9))
for ax, col in zip(axes.flat, CONTINUOUS):
    sns.boxplot(data=df_clean, x=TARGET, y=col, hue=TARGET, order=GENRES, hue_order=GENRES,
                palette=GENRE_COLORS, showfliers=False, legend=False, ax=ax)
    ax.set_title(col)
    ax.set_xlabel("")
    ax.set_ylabel("")
fig.suptitle("Audio vlastnosti podle žánru", fontsize=16)
plt.tight_layout()
plt.show()
```

Mediány v tabulce – barva v každém řádku ukazuje, který žánr má danou vlastnost nejnižší (modrá) a nejvyšší (červená):

```python
genre_medians = df_clean.groupby(TARGET)[CONTINUOUS].median().T
genre_medians.style.background_gradient(axis=1, cmap="coolwarm").format(precision=3)
```

Kategorické vlastnosti `mode` a `key` zobrazíme jako podíly:

```python
KEY_NAMES = ["C", "C♯", "D", "D♯", "E", "F", "F♯", "G", "G♯", "A", "A♯", "H"]

fig, axes = plt.subplots(1, 2, figsize=(20, 5), gridspec_kw={"width_ratios": [1, 2.5]})

mode_share = df_clean.groupby(TARGET)["mode"].mean().reindex(GENRES) * 100
axes[0].bar(GENRES, mode_share, color=[GENRE_COLORS[g] for g in GENRES])
axes[0].axhline(50, color="gray", linestyle="--", linewidth=1)
axes[0].set_ylim(0, 100)
axes[0].set_title("Podíl písní v dur [%]")

key_share = pd.crosstab(df_clean[TARGET], df_clean["key"], normalize="index") * 100
key_share.columns = KEY_NAMES
sns.heatmap(key_share, annot=True, fmt=".0f", cmap="Blues", cbar=False, ax=axes[1])
axes[1].set_title("Zastoupení tónin v jednotlivých žánrech [%]")
axes[1].set_xlabel("Tónina")
axes[1].set_ylabel("")

plt.tight_layout()
plt.show()
```

**Profily žánrů podle EDA:**

## 5️⃣ Rozdělení dat na X a y, trénovací a testovací množinu

- **`y`** (výstup) = `playlist_genre` – žánr, který chceme předpovídat.
- **`X`** (vstupy) = **11 audio vlastností**. Oproti EDA vynecháváme:

  - `track_popularity` – nepopisuje zvuk písně (viz závěr EDA).

Data dělíme v poměru **80 % trénovací / 20 % testovací**:
- `stratify=y` zajistí, že poměr žánrů bude v obou částech stejný,
- `random_state=42` zajistí, že rozdělení bude při každém spuštění stejné (výsledky jsou opakovatelné).

Testovací data jsou „zamčená“ – použijeme je **jen jednou, na úplném konci**, k ověření, jak si modely vedou na písních, které nikdy neviděly. Hyperparametry ladíme pouze na trénovacích datech pomocí křížové validace.

```python
from sklearn.model_selection import train_test_split

FEATURES = [c for c in AUDIO_FEATURES if c != "key"]

X = df_clean[FEATURES]
y = df_clean[TARGET]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print(f"Vstupní vlastnosti ({len(FEATURES)}): {FEATURES}")
print(f"Trénovací množina: {len(X_train)} písní")
print(f"Testovací množina: {len(X_test)} písní")
```

Kontrola, že stratifikace funguje – zastoupení žánrů je v obou částech prakticky stejné:

```python
pd.DataFrame({
    "trénovací [%]": y_train.value_counts(normalize=True) * 100,
    "testovací [%]": y_test.value_counts(normalize=True) * 100,
}).reindex(GENRES).round(1)
```

## 6️⃣ Vytvoření ML modelů

| Model | Role | Proč |
|---|---|---|
| `DummyClassifier` | **baseline** | vždy tipne nejčastější žánr – spodní hranice, kterou musí každý užitečný model překonat |
| `DecisionTreeClassifier` | **referenční model** | jednoduchý a dobře interpretovatelný, nepotřebuje škálování dat |
| `SVC` (SVM s RBF jádrem) | **hlavní model** | umí vytvořit nelineární hranice mezi žánry, které se v datech překrývají |

**Ladění hyperparametrů:** pro každý model vyzkoušíme více kombinací hyperparametrů pomocí `GridSearchCV`. Ten každou kombinaci ohodnotí **křížovou validací (5-fold)**: trénovací data rozdělí na 5 částí, model 5× natrénuje na 4 částech a ověří na zbylé páté, a výsledné přesnosti zprůměruje. Díky tomu vybíráme hyperparametry bez použití testovacích dat.

### 6.1 Referenční model – rozhodovací strom

Nejdůležitějším hyperparametrem stromu je **`max_depth`** (maximální hloubka) – určuje, jak složitá pravidla se strom může naučit:
- příliš mělký strom je moc jednoduchý a nezachytí rozdíly mezi žánry (*underfitting*),
- příliš hluboký strom si trénovací data „zapamatuje“ i se šumem a na nových datech selhává (*overfitting*).

Nejdřív se podíváme, jak hloubka ovlivňuje přesnost na trénovacích datech a při křížové validaci:

```python
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier

depths = range(1, 21)
train_acc, cv_acc = [], []
for depth in depths:
    tree = DecisionTreeClassifier(max_depth=depth, random_state=42)
    cv_acc.append(cross_val_score(tree, X_train, y_train, cv=5).mean())
    train_acc.append(tree.fit(X_train, y_train).score(X_train, y_train))

best_depth = depths[int(np.argmax(cv_acc))]

plt.figure(figsize=(10, 5))
plt.plot(depths, train_acc, marker="o", label="trénovací data")
plt.plot(depths, cv_acc, marker="o", label="křížová validace (5-fold)")
plt.axvline(best_depth, color="gray", linestyle="--", label=f"nejlepší hloubka = {best_depth}")
plt.xlabel("max_depth (maximální hloubka stromu)")
plt.ylabel("Přesnost")
plt.title("Rozhodovací strom: vliv hloubky na přesnost")
plt.xticks(list(depths))
plt.legend()
plt.show()
```

Graf ukazuje typický průběh **přeučení**: na trénovacích datech přesnost s hloubkou roste téměř ke 100 % (98 % při hloubce 20 – strom se data učí nazpaměť), ale přesnost při křížové validaci roste jen do hloubky 9 (necelých 48 %) a pak naopak klesá.

Kromě hloubky vyzkoušíme v `GridSearchCV` ještě:
- **`min_samples_leaf`** – minimální počet písní v listu stromu; vyšší hodnota brání vzniku listů, které popisují jen pár konkrétních písní,
- **`criterion`** – způsob měření „čistoty“ rozdělení (`gini` nebo `entropy`).

```python
from sklearn.model_selection import GridSearchCV

tree_grid = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_grid={
        "max_depth": [6, 8, 10, 12, 15],
        "min_samples_leaf": [1, 10, 30, 60, 100],
        "criterion": ["gini", "entropy"],
    },
    cv=5,
    n_jobs=-1,
)
tree_grid.fit(X_train, y_train)

print("Nejlepší hyperparametry:", tree_grid.best_params_)
print(f"Přesnost při křížové validaci: {tree_grid.best_score_:.1%}")
```

Pět nejlepších kombinací:

```python
tree_results = pd.DataFrame(tree_grid.cv_results_)
tree_results = tree_results[["param_criterion", "param_max_depth", "param_min_samples_leaf", "mean_test_score"]]
tree_results.columns = ["criterion", "max_depth", "min_samples_leaf", "přesnost (CV)"]
tree_results.sort_values("přesnost (CV)", ascending=False).head(5).round(4)
```

- Nejlepší strom má **hloubku 12, alespoň 60 písní v každém listu a kritérium `gini`** → přesnost při křížové validaci **48,4 %**.
- Omezení velikosti listů (`min_samples_leaf` 30–60) pomáhá: strom pak může být hlubší (10–15), aniž by se přeučil, a je přesnější než strom omezený jen hloubkou (necelých 48 % v grafu výše).
- Rozdíly mezi nejlepšími kombinacemi jsou jen desetiny procenta → výsledek je stabilní a na přesném nastavení příliš nezáleží.

### 6.2 Hlavní model – SVM (Support Vector Machine)

SVM hledá mezi třídami hranici s co největším odstupem (*margin*). S **RBF jádrem** může být hranice nelineární, což se hodí, protože se žánry v prostoru vlastností překrývají. Šest tříd SVM řeší porovnáváním dvojic žánrů (*one-vs-one*).

Ladíme dva hyperparametry:
- **`C`** – jak přísně model trestá chyby na trénovacích datech. Velké `C` = hranice se snaží vyhovět každé písni (riziko přeučení), malé `C` = hladší, obecnější hranice.
- **`gamma`** – jak daleko sahá vliv jedné trénovací písně. Velká `gamma` = velmi „zvlněná“ hranice kolem jednotlivých bodů (přeučení), malá `gamma` = hladká hranice.

⚠️ **Dvě praktické poznámky:**
- SVM počítá se vzdálenostmi, proto vyžaduje **standardizovaná data** (EDA ukázala velmi rozdílná měřítka). `StandardScaler` je spojený s modelem do `Pipeline`, takže se při křížové validaci škálování vždy naučí jen z trénovací části – nedochází k úniku informací z validačních dat.
- Trénování SVM je výpočetně náročné (čas roste zhruba s druhou mocninou počtu dat). Hyperparametry proto ladíme na **náhodném vzorku 8 000 písní** z trénovací množiny a výsledný model pak natrénujeme na celé trénovací množině. Buňka trvá v Colabu zhruba 1–2 minuty.

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

svm_sample = X_train.sample(8000, random_state=42)

svm_grid = GridSearchCV(
    make_pipeline(StandardScaler(), SVC(kernel="rbf")),
    param_grid={
        "svc__C": [0.1, 1, 10, 100],
        "svc__gamma": [0.01, 0.1, 1],
    },
    cv=5,
    n_jobs=-1,
)
svm_grid.fit(svm_sample, y_train.loc[svm_sample.index])

print("Nejlepší hyperparametry:", svm_grid.best_params_)
print(f"Přesnost při křížové validaci: {svm_grid.best_score_:.1%}")
```

```python
svm_results = pd.DataFrame(svm_grid.cv_results_).pivot(
    index="param_svc__C", columns="param_svc__gamma", values="mean_test_score"
)

plt.figure(figsize=(7, 5))
sns.heatmap(svm_results, annot=True, fmt=".3f", cmap="viridis")
plt.title("SVM: přesnost při křížové validaci")
plt.xlabel("gamma")
plt.ylabel("C")
plt.show()
```

- Nejlepší kombinace je **`C = 1`, `gamma = 0.1`** s přesností **52,4 %** při křížové validaci. SVM je tak už na menším vzorku dat lepší než nejlepší rozhodovací strom (48,4 %).
- **Příliš velká `gamma` (1)** dává ve všech případech výrazně horší výsledky: hranice se „obtáčí“ kolem jednotlivých písní a model nezobecňuje. Při malém `C = 0.1` je výsledek dokonce jen 23 % – skoro jako hádání.
- Oba parametry spolu souvisí: s rostoucím `C` je potřeba menší `gamma` (např. `C = 100`, `gamma = 0.01` dosahuje téměř stejných 52,1 %). Oba totiž ovlivňují, jak „složitou“ hranici model vytvoří.

## 7️⃣ Trénování a testování modelů

Všechny tři modely teď natrénujeme s nejlepšími nalezenými hyperparametry **na celé trénovací množině** (SVM tedy už ne jen na vzorku 8 000 písní) a poprvé je ověříme na **testovacích datech**.

Porovnání přesnosti na trénovacích a testovacích datech zároveň ukáže, zda se model nepřeučil – velký rozdíl by znamenal, že si model trénovací data spíš pamatuje, než aby se naučil obecná pravidla.

```python
import time

from sklearn.dummy import DummyClassifier

models = {
    "Baseline (Dummy)": DummyClassifier(strategy="most_frequent"),
    "Rozhodovací strom": tree_grid.best_estimator_,
    "SVM (RBF)": svm_grid.best_estimator_,
}

results = []
for name, model in models.items():
    start = time.perf_counter()
    model.fit(X_train, y_train)
    train_time = time.perf_counter() - start

    results.append({
        "model": name,
        "přesnost – trénovací": model.score(X_train, y_train),
        "přesnost – testovací": model.score(X_test, y_test),
        "čas trénování [s]": train_time,
    })

results = pd.DataFrame(results).set_index("model")
results.round(3)
```

Podrobnější výsledky nejlepšího modelu (SVM) pro jednotlivé žánry:
- **precision** – z písní, které model označil za daný žánr, kolik jich do něj opravdu patří,
- **recall** – kolik písní daného žánru model správně našel,
- **f1-score** – harmonický průměr precision a recall.

```python
from sklearn.metrics import classification_report

y_pred_svm = models["SVM (RBF)"].predict(X_test)
print(classification_report(y_test, y_pred_svm, digits=3))
```

**Shrnutí výsledků:**
- **SVM je nejlepší model** s přesností **53,9 %** na testovacích datech, rozhodovací strom dosáhl **49,0 %**. Oba jsou výrazně lepší než baseline (**19,4 %**) – SVM určí správně zhruba 2,8× víc písní než tipování nejčastějšího žánru.
- Testovací přesnost odpovídá výsledkům křížové validace (strom 48,4 % → 49,0 %). U SVM je dokonce vyšší (52,4 % → 53,9 %), protože finální model se trénoval na celé trénovací množině (19 472 písní), ne jen na vzorku 8 000 písní.
- Rozdíl mezi trénovací a testovací přesností je u obou modelů malý (4–5 procentních bodů) → modely **nejsou výrazně přeučené**.
- Za vyšší přesnost SVM „platí“ řádově delším trénováním než strom (viz sloupec *čas trénování*).
- Výsledky podle žánrů potvrzují očekávání z EDA: nejlépe model rozpoznává **EDM, rock a rap** (f1 ≈ 0,63–0,66), nejhůře **pop** (f1 = 0,31) a **latin** (f1 = 0,39), které nemají výraznou zvukovou „signaturu“.

Přesnost kolem 54 % se může zdát nízká, ale je potřeba ji vidět v kontextu: žánr se určuje jen z 11 čísel popisujících zvuk, žánry se navzájem prolínají a štítek pochází z názvu playlistu, ne od hudebního odborníka.

### 8.3 Úspěšnost podle žánrů – strom vs. SVM

Pro porovnání použijeme **F1-score** (kombinace precision a recall) – čím vyšší, tím lépe model daný žánr rozpoznává.

```python
from sklearn.metrics import f1_score

y_pred_tree = models["Rozhodovací strom"].predict(X_test)

f1_by_genre = pd.DataFrame({
    "Rozhodovací strom": f1_score(y_test, y_pred_tree, labels=GENRES, average=None),
    "SVM (RBF)": f1_score(y_test, y_pred_svm, labels=GENRES, average=None),
}, index=GENRES).sort_values("SVM (RBF)", ascending=False)

ax = f1_by_genre.plot(kind="bar", figsize=(10, 5), rot=0, color=["#fdae6b", "#3182bd"])
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f", padding=2)
ax.set_ylim(0, 0.8)
ax.set_ylabel("F1-score")
ax.set_xlabel("")
ax.set_title("F1-score podle žánru (testovací data)")
plt.show()
```

- Oba modely mají stejnou trojici nejlépe rozpoznávaných žánrů (EDM, rap, rock) i stejné nejslabší žánry (latin, pop). Obtížnost jednotlivých žánrů je tedy daná hlavně **daty** (tím, jak se žánry zvukově překrývají), ne volbou modelu.

### 8.4 Které vlastnosti modely využívají

- **Rozhodovací strom** – `feature_importances_` udává, jak velký podíl na rozhodování stromu má každá vlastnost.
- **SVM** žádnou takovou hodnotu nemá, proto použijeme **permutation importance**: hodnoty jedné vlastnosti v testovacích datech náhodně zamícháme a změříme, o kolik klesne přesnost. Čím větší pokles, tím víc model na vlastnosti závisí. Výpočet je pomalý (model musí znovu předpovídat), proto ho děláme na vzorku 1 000 testovacích písní – buňka trvá v Colabu zhruba minutu.

```python
from sklearn.inspection import permutation_importance

tree_importance = pd.Series(models["Rozhodovací strom"].feature_importances_, index=FEATURES)

perm_sample = X_test.sample(1000, random_state=42)
perm = permutation_importance(
    models["SVM (RBF)"], perm_sample, y_test.loc[perm_sample.index],
    n_repeats=5, random_state=42, n_jobs=-1,
)
svm_importance = pd.Series(perm.importances_mean, index=FEATURES)

fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))
tree_importance.sort_values().plot(kind="barh", ax=axes[0], color="#fdae6b")
axes[0].set_title("Rozhodovací strom – feature importance")
axes[0].set_xlabel("Podíl na rozhodování stromu")
svm_importance.sort_values().plot(kind="barh", ax=axes[1], color="#3182bd")
axes[1].set_title("SVM – permutation importance")
axes[1].set_xlabel("Pokles přesnosti po zamíchání vlastnosti")
plt.tight_layout()
plt.show()
```

- Oba modely se shodují, že mezi **nejdůležitější** vlastnosti patří `danceability`, `speechiness` a `energy`, a že **`liveness` a `mode` nemají téměř žádný vliv** – přesně jak naznačila EDA.
- **Strom** nejvíc využívá `tempo`: hned pod sebou má pravidla „tempo ≤ 121 BPM“ a „tempo ≤ 128 BPM“, která vymezují přesně ustálené tempo EDM, které ukázala EDA. U SVM je `tempo` až na 5. místě.
- **SVM** nejvíc závisí na `danceability` – po jejím zamíchání klesne přesnost asi o 12 procentních bodů.

### 8.5 Jak rozhoduje rozhodovací strom

Výhodou stromu je, že jeho pravidla lze přímo přečíst. Celý strom má hloubku 12, pro přehlednost zobrazíme jen horní 3 úrovně. V každém uzlu je pravidlo, podíl písní, které do uzlu došly (`samples`), zastoupení žánrů (`value`, v pořadí edm, latin, pop, r&b, rap, rock) a převažující žánr (`class`).

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(24, 10))
plot_tree(
    models["Rozhodovací strom"], max_depth=2, feature_names=FEATURES,
    class_names=list(models["Rozhodovací strom"].classes_), filled=True,
    impurity=False, proportion=True, rounded=True, precision=2, fontsize=11,
)
plt.title("Horní 3 úrovně rozhodovacího stromu", fontsize=16)
plt.show()
```

Pravidla stromu odpovídají zjištěním z EDA:
- **Kořen** dělí písně podle `speechiness` ≤ 0,14. Ve větvi s vyšším podílem mluveného slova (vpravo, 25 % písní) je **46 % rapu** – oproti 19 % v celých datech.
- Vlevo (málo mluveného slova) pravidlo `danceability` ≤ 0,56 oddělí **méně taneční písně, mezi kterými je 43 % rocku**.
- Taneční písně bez rapu se dále dělí podle **tempa** (≤ 121 BPM) a hned pod tím podle tempa ≤ 128 BPM – tato dvě pravidla vymezují ustálené tempo **EDM**.
- Barva uzlu odpovídá převažujícímu žánru a její sytost tomu, jak výrazně převažuje. Ani po třech úrovních není žádný uzel „čistý“ – proto strom potřebuje mnohem víc úrovní a i tak dosahuje jen ~49 %.
