## 🧩 Posílené učení (Reinforcement Learning)
- inspirováno v **psychlogii a neurovědě**
- kopíruje **vzor jakým se učí člověk/zvíře**, nebo biologický vzory jakými mozek reaguje na podněty
- velmi **zobecněný způsob učení**


**Hlavní princip:** systém odměn a trestů 🟢🔴
1. Počáteční stav
2. Akce
3. Vyhodnocení akce
4. Odměna / trest
5. Nový stav
6. Zapomatování si následků akce

### 2. Inicializace virtuálního prostředí
- programovat budeme v rámci `virtual environment`
- to už je vytvořené, ale nutné jej **inicializovat**¨

**Vytvoření Virtual Environment:**
- Stiskněte `Ctrl+Shift+P`
- zadejte `Python: Create Environment`
- vyberte `Venv`
- vyberte interpreter `Python`

**Aktivace Virtual Environment:**
- pokud se už tak nestalo automaticky otevřete pomocí `Ctrl+Shift+;` terminál a zadejte tyto příkazy:
  -  `cd .venv`
  - `Set-ExecutionPolicy -ExecutionPolicy Unrestricted -Scope Process`
  - `.\Scripts\Activate.ps1` nebo `.\Scripts\activate`

- pokud je aktivace úspěšná, budete mít v terminálu prefix `(.venv) `🟢

activate_venv.png
