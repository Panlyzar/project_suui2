### 🔷 Matplotlib - grafy a vizualizace výsledků
Tato šikovná kniovna umí vizulizovat téměř cokoliv, její nevýhoda je ovšem vcelku neintuitivní používání, zejména pro začátečníky. Proto si pojďme ukázat pár jednoduchých příkladů.

````python
import matplotlib.pyplot as plt
````

````python
# Testovací data - jednorozměrná
x = np.array([1,2,3,4,5])
y = np.array([0.2,0.25,0.33,0.15,0.10])
````

````python
# Vytvoření a vykreslení grafu

# Funce plot vykreslí průběh
# b = bolt line
plt.plot(x, y, 'b', label = 'Testing data')

# Název grafu, os a další parametry
plt.title('My testing plot')
plt.xlabel('x')
plt.ylabel('y')

# Zobraz legendu
plt.legend()

# SAve
plt.savefig("test.png")

# Vykresli
plt.show()
````

````python
# Vyčištění vykreslovací paměti
plt.clf()
````

````python
# Místo čáry mohu použít pouze tečky - body

# bo - blue dot
plt.plot(x, y, 'bo', label = 'Testing data')
plt.title('My testing plot')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
#show
plt.show()
````

````python
# Přidám další dimenzi - z
z = np.array([0.5, 0.6, 0.7, 0.2, 0.1])
````

````python
# Funce plot vykreslí průběh pro x,y a x,z
plt.plot(x, y, 'b', label = 'Testing data')
plt.plot(x, z, 'bo', label = 'Other data')

plt.title('My testing plot')
plt.xlabel('x')
plt.ylabel('y')

#Přidám vykreslení legendy
plt.legend()

# zobraz
plt.show()
````

````python
# Matplotlib umí i 3D
fig = plt.figure()
ax = plt.axes(projection='3d')

# Vytvořím si 3-dimezionální data x,y,z pomocí numpy, a provedu na nich kalkulaci sin a cos
zline = np.linspace(0, 15, 1000)
ampl = np.linspace(1, 10, 1000)
xline = np.cos(zline) * ampl
yline = np.sin(zline) * ampl

# Vykreslím spirálu
ax.plot3D(xline, yline, zline, 'gray')

plt.show()
````

````python
plt.figure()
plt.plot(xline, yline, color="gray")

plt.show()
````

````python
plt.figure()
plt.plot(xline, label="cos")
plt.plot(yline, label="sin")

# Dont show horizontal grid-lines
plt.grid(axis='y', linestyle='--')

plt.legend()

plt.show()
````

````python
# A generuje i šikovné heatmapy
heat_data = np.random.rand(5, 5)
heat_data
````
